```python
# Activity 2: Deadline & Date Formatter
# Input Data from user
#version 2
from datetime import datetime

event_title = input("Enter event title: ").strip()
year = int(input("Enter year:"))
month = input("Enter month:").strip()
day = int(input("Enter day:"))

if month.isdigit():
    month = int(month)
else:
    month = datetime.strptime(month.title(), "%B").month

# Create datetime object
event_date = datetime(year, month, day)

formatted_event = event_title.title()
formatted_month = event_date.strftime("%B")
formatted_date = event_date.strftime("%B %d, %Y")
weekday = event_date.strftime("%A")

print("--- Formatted Output ---")
print(f"Event: {formatted_event.upper()}")
print(f"Date: {formatted_date}")
print(f"Year: {year}")
print(f"Month: {formatted_month}")
print(f"Day: {day}")
print(f"Weekday: {weekday}")
```
