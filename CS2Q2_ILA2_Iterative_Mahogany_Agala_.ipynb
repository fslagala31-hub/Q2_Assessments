```python
secret_number = 17
tries = 0
max_tries = 5
won = False

print("=== Guess the number! (*^▽^*)")
print("Guess the secret number! You only have 5 valid tries.")

while tries < max_tries:
  user_input = input("\nEnter your guess: ")
  guess = int(user_input)

  if guess < 0:
    print("Opps! Negative numbers don't count, try again.")
    continue

  tries = tries + 1
  tries_left = max_tries - tries

  if guess == secret_number:
    print("CONGRATULATIONS! You found the secret numer! (●ˇ∀ˇ●)")
    won = True
    break
  elif guess < secret_number:
    print("Too low! Tries left:" + str(tries_left))
  else:
    print("Too high! Tries left:" + str(tries_left))

if won == False:
  print("\nGame over! Better luck next time(┬┬﹏┬┬)")
```
