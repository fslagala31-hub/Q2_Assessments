```python
#Activity 3 (Graded or Non-Graded Formative Assessment): Data Analysis Mini App

scores = [150, 200, 250, 300, 350]

def show_scores(my_list):
  print("\n---Current scores (≧∇≦)---")
  for i in range(len(my_list)):
    print("Index", i, ":", my_list[i])

def find_score(my_list, target):
  if target in my_list:
    print("YAY ╰(*°▽°*)╯! Found score", target, "in the list.")
  else:
    print("Score", target, "was not found ＞︿＜.")

def show_stats(my_list):
  total = sum(my_list )
  average = total / len(my_list)
  print("\n--- Score stats---")
  print("Total point:", total)
  print("Average score:", average)
  print("Highest score:", max(my_list))
  print("Lowest score:", min(my_list))

def change_score(my_list, index, new_value):
  my_list[index] = new_value
  print("Update index", index, "to", new_value)

print("=== Game score tracker ===")

show_scores(scores)
show_stats(scores)

print("\n--- Searching for a score---")
find_score(scores, 500)

print("\n--- Updating a score ---")
change_score(scores, 2, 250)

show_scores(scores)
show_stats(scores)

```
